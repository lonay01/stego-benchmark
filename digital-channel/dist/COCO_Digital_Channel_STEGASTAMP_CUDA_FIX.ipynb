## 2. Restore the reviewed system dependencies

Only missing MagickWand/GL libraries are installed; no method is changed.


```python
# Only RoSteALS/SepWater need these author-import system libraries.
# Restore the setup used by RoSteALS_AUTHOR_FIX and RoSteALS_SepWater_RECOVERY.
# No model source, checkpoint, Python lock, payload or matching rule is changed.
import ctypes
import ctypes.util

def ensure_author_system_dependencies(models, in_colab):
    affected = {"rosteals", "sepwater"}
    selected = affected if models == "all" else affected.intersection(models)
    if not selected:
        return {}

    requirements = {
        "libmagickwand-dev": (
            ("MagickWand-6.Q16", "MagickWand-6.Q16HDRI", "MagickWand-7.Q16HDRI",
             "MagickWand-7.Q16", "MagickWand"), "MagickWandGenesis"),
        "libgl1": (("GL",), None),
    }

    def inspect_libraries():
        found = {}
        for package, (names, symbol) in requirements.items():
            for name in names:
                path = ctypes.util.find_library(name)
                if not path:
                    continue
                try:
                    library = ctypes.CDLL(path)
                    if symbol:
                        getattr(library, symbol)
                except (OSError, AttributeError):
                    continue
                found[package] = path
                break
        return found

    found = inspect_libraries()
    missing = [package for package in requirements if package not in found]
    if missing:
        install = ["apt-get", "-qq", "install", "-y", "--no-install-recommends", *missing]
        if not in_colab or not shutil.which("apt-get") or os.geteuid() != 0:
            raise RuntimeError(
                "RoSteALS/SepWater system libraries are missing. On Debian/Ubuntu run: "
                "sudo apt-get update && sudo " + " ".join(install)
                + ". Then rerun this setup cell; keep the existing output directory.")
        print("Installing missing system libraries for " + ", ".join(sorted(selected))
              + ": " + ", ".join(missing), flush=True)
        subprocess.run(["apt-get", "-qq", "update"], check=True, timeout=600)
        subprocess.run(install, check=True, timeout=600)
        found = inspect_libraries()
        missing = [package for package in requirements if package not in found]
        if missing:
            raise RuntimeError("System-library preflight failed after installation: "
                               + ", ".join(missing) + ". No model has been started.")
    print("RoSteALS/SepWater system-library preflight passed:", found, flush=True)
    return found

AUTHOR_SYSTEM_LIBRARIES = ensure_author_system_dependencies(MODELS, IN_COLAB)
```

## StegaStamp Clean replay repair

Only the StegaStamp Clean comparison is changed. No decoder, weights, payloads,
PSNR matching, attacks or other methods are changed. The original source ZIP
and run identity stay the same; completed cases are reused.

For StegaStamp Matched Quality only, a **single changed bit out of 100** may be
accepted when both extractions are non-exact and two additional decodes match
the **first fresh decode exactly**. No voting, best-BER selection or bit repair.
All other Clean mismatches still stop execution. A partial case cannot switch
its Clean reference. Model, dependency and PNG checksum checks remain enabled.

The old bank result is retained. Attack deltas use the actual fresh Clean result.
The discrepancy and all verification decodes are saved under
`watermarking_digital_channel_v1/stegastamp/clean-replay/` and referenced in
the case record. This is an explicit replay-policy exception, not a claim that
the historical and current outputs are identical.

**Run all cells with the same bank/output paths and N_IMAGES=5.** No new ZIP is
needed if the original source ZIP was saved on Drive. If requested, upload the
same `COCO_Digital_Channel_RESUMABLE_source.zip` as before. Do not delete results.


```python
# Narrow StegaStamp repair; the verified original release is NOT edited.
HOTFIX_SOURCE = r'''
"""Scoped Clean replay handling for the frozen digital-channel release.

No decoder, weight, attack, saved carrier or historical measurement is changed.
Only StegaStamp Matched Quality may accept a stable, single-bit replay drift.
The FIRST fresh extraction is the paired Clean control, never the best retry.
"""

import ast
import hashlib
from pathlib import Path
import sys
import time

VERSION = "stegastamp-clean-replay-v1"
METHOD = "stegastamp"
WORKER_SHA = "57f68d4e9c6134c1f5264703e6e00ac1a6ef55a0040265f5b0c6dfd46ef6cf2e"
FIELDS = ("status", "decoded_bits", "ber", "exact")
REPEATS = 2  # Verification only; neither voting nor selection by BER.


def same_result(left, right):
    return all(left.get(key) == right.get(key) for key in FIELDS)


def drift_positions(case, fresh):
    """A deliberately bounded exception, not a general BER tolerance."""
    bank = case["clean"]
    if case["method"] != METHOD or case["mode"] not in {"matched_39", "matched_42", "matched_45"}:
        return None
    capacity = case["assignment"]["capacity"]
    if capacity["information_bits"] != 100 or capacity["information_indices"] != list(range(100)):
        return None
    expected = case["assignment"]["payload"]
    if len(expected) != 100 or any(bit not in (0, 1) for bit in expected):
        return None
    for row in (bank, fresh):
        bits = row.get("decoded_bits")
        if (row.get("status") != "measured" or row.get("exact") is not False
                or not isinstance(bits, list) or len(bits) != 100
                or any(type(bit) is not int or bit not in (0, 1) for bit in bits)
                or row.get("information_bits") != 100):
            return None
        errors = sum(x != y for x, y in zip(bits, expected))
        if row.get("ber") != errors / 100 or row.get("information_errors") != errors:
            return None
        if row.get("details", {}).get("implementation") != "wibe_onnx":
            return None
    changed = [i for i, (a, b) in enumerate(zip(bank["decoded_bits"], fresh["decoded_bits"])) if a != b]
    return changed if len(changed) == 1 else None


def check_clean(clean, case, adapter, metadata, rng_seed, root, execution, prior, job):
    """Keep original strict checks except the audited StegaStamp replay case."""
    from wmbench.io import read_json, write_json, sha256
    from wmbench_full_payload.storage import EvidenceError, inside
    from wmbench_digital import worker

    root = Path(root)
    name = case["method"]
    historical = case["clean"]
    # A partially decoded carrier must never combine two different Clean controls.
    previous_clean = prior.get("clean")
    if name == METHOD and previous_clean and not same_result(previous_clean, clean):
        write_json(root / name / "clean-replay" / f"resume-conflict-{case['id']}-{execution['session']}.json",
                   {"fix": VERSION, "previous_clean": previous_clean, "fresh": clean,
                    "execution": execution, "accepted": False})
        raise EvidenceError("StegaStamp Clean changed within a partial case; existing attack rows retained")
    if same_result(clean, historical):
        return

    # Preserve the same diagnostic and hard-stop behavior for every other failure.
    diagnostic_name = (case["id"] + "-" + execution["session"] if name == METHOD else case["id"]) + ".json"
    write_json(root / name / "clean-drift" / diagnostic_name,
               {"fresh": clean, "bank": historical, "execution": execution})
    changed = drift_positions(case, clean)
    if changed is None:
        raise EvidenceError("Fresh Clean differs from saved bank control; no attack scores accepted")

    carrier = inside(Path(job["bank"]), case["output"])
    if sha256(carrier) != case["carrier_sha256"]:
        raise EvidenceError("StegaStamp carrier checksum changed during Clean replay")
    repeats = []
    for _ in range(REPEATS):
        if time.time() >= job["deadline"]:
            raise TimeoutError("Session budget reached during StegaStamp Clean verification")
        repeats.append(worker.measured(adapter, carrier, metadata, case, rng_seed))
    stable = all(same_result(clean, row) for row in repeats)
    relative = Path(name) / "clean-replay" / f"{case['id']}-{execution['session']}.json"
    receipt = {"fix": VERSION, "case_key": case["key"], "carrier_sha256": case["carrier_sha256"],
               "bank": historical, "first_fresh": dict(clean), "verification_repeats": repeats,
               "changed_indices_zero_based": changed, "execution": execution, "accepted": stable,
               "rule": "matched-only; exactly one bit; both non-exact; two identical verification extractions",
               "paired_control": "first_fresh", "decoder_changed": False, "bank_changed": False}
    write_json(root / relative, receipt)
    if read_json(root / relative) != receipt:
        raise EvidenceError("StegaStamp replay receipt did not persist")
    if not stable:
        raise EvidenceError("StegaStamp Clean is unstable on repeat; no attack scores accepted")
    clean["clean_replay"] = {"fix": VERSION, "receipt": relative.as_posix(),
                            "bank_ber": historical["ber"], "changed_indices_zero_based": changed,
                            "paired_control": "first_fresh"}
    print(f"STEGASTAMP CLEAN REPLAY {case['id']}: one bit differs; "
          f"bank BER={historical['ber']:.2%}, first fresh BER={clean['ber']:.2%}; "
          f"2/2 repeats stable. Historical bank retained. Audit: {root / relative}", flush=True)


def install_worker_patch():
    """Replace one comparison block in memory; frozen files remain byte-identical."""
    from wmbench_digital import worker

    source = Path(worker.__file__).read_bytes()
    if hashlib.sha256(source).hexdigest() != WORKER_SHA:
        raise RuntimeError("Unexpected worker version; refusing to apply the StegaStamp-only fix")
    tree = ast.parse(source.decode("utf-8"))
    function = next(n for n in tree.body if isinstance(n, ast.FunctionDef) and n.name == "decode")
    matches = 0

    class ReplaceGate(ast.NodeTransformer):
        def visit_For(self, node):
            nonlocal matches
            if (isinstance(node.target, ast.Name) and node.target.id == "field"
                    and isinstance(node.iter, ast.Tuple)
                    and [getattr(e, "value", None) for e in node.iter.elts] == list(FIELDS)):
                matches += 1
                replacement = ast.parse("_check_stegastamp_clean(clean, case, adapter, metadata, "
                                        "rng_seed, root, execution, prior, job)").body[0]
                return ast.copy_location(replacement, node)
            return self.generic_visit(node)

    function = ReplaceGate().visit(function)
    if matches != 1:
        raise RuntimeError("Expected exactly one Clean comparison block")
    module = ast.fix_missing_locations(ast.Module(body=[function], type_ignores=[]))
    worker.__dict__["_check_stegastamp_clean"] = check_clean
    exec(compile(module, str(worker.__file__) + " [StegaStamp replay fix]", "exec"), worker.__dict__)


def redirected_command(command, env):
    """Only the isolated StegaStamp decode child uses this repair entry point."""
    import os
    from wmbench.io import read_json

    command = list(command)
    marker = ["-m", "wmbench_digital.runner", "decode", "--settings"]
    for i in range(len(command) - len(marker)):
        if command[i:i+len(marker)] != marker:
            continue
        job = read_json(command[i+len(marker)])
        if job["model"]["id"] != METHOD:
            return command, env
        command[i+1] = "stegastamp_clean_hotfix"
        env = dict(env)
        env["PYTHONPATH"] = str(Path(__file__).resolve().parent) + os.pathsep + env.get("PYTHONPATH", "")
        return command, env
    return command, env


def main():
    from wmbench_digital import runner
    from wmbench.io import read_json

    action = sys.argv[1] if len(sys.argv) > 1 else ""
    if action == "decode":
        job = read_json(Path(sys.argv[sys.argv.index("--settings")+1]))
        if job["model"]["id"] != METHOD:
            raise RuntimeError("This decode entry point is only for StegaStamp")
        install_worker_patch()
    elif action == "run":
        from wmbench import colab_canary
        original = colab_canary.monitored_process

        def monitored(command, env, *args, **kwargs):
            command, env = redirected_command(command, env)
            return original(command, env, *args, **kwargs)

        colab_canary.monitored_process = monitored
    else:
        raise RuntimeError("Use the original runner for validate/report; repair supports run/decode only")
    runner.main()


if __name__ == "__main__":
    main()
'''
HOTFIX_SOURCE = HOTFIX_SOURCE.lstrip("\n")
HOTFIX_SHA = "60fc74e04651d01346029cb5df21cad61cf1e889069d71d86eed396fc6317842"
if hashlib.sha256(HOTFIX_SOURCE.encode("utf-8")).hexdigest() != HOTFIX_SHA:
    raise ValueError("StegaStamp repair source changed")
HOTFIX_DIR = SCRATCH_ROOT / "repairs/stegastamp-clean-replay-v1"
HOTFIX_DIR.mkdir(parents=True, exist_ok=True)
for folder in (HOTFIX_DIR, OUTPUT_ROOT / "_release/repairs/stegastamp-clean-replay-v1"):
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / "stegastamp_clean_hotfix.py"
    if target.exists() and target.read_bytes() != HOTFIX_SOURCE.encode("utf-8"):
        raise ValueError(f"Different repair already exists: {target}")
    if not target.exists():
        target.write_bytes(HOTFIX_SOURCE.encode("utf-8"))
    if hashlib.sha256(target.read_bytes()).hexdigest() != HOTFIX_SHA:
        raise ValueError("Repair readback failed")
ENV["PYTHONPATH"] = str(HOTFIX_DIR) + os.pathsep + ENV["PYTHONPATH"]
print("StegaStamp-only Clean replay repair enabled; bank and completed results retained.")
```

## Resume after temporary cache loss

This fixes `FileNotFoundError: .../attack-cache/.../owner.json` during cleanup
of a completed method after Colab loses its temporary disk. Saved results are
still checked before cleanup. An absent ZIP is skipped; an existing ZIP still
requires the original ownership, path and checksum checks.

Decoders, attacks, metrics and the previous StegaStamp repair are unchanged.
Keep the same bank/output paths and `N_IMAGES = 5`; run all cells in order.
The original source ZIP is reused from Drive. Do not delete results or cache
metadata manually. This repair does not regenerate finished measurements.


```python
# Resume repair only: no changes to decoders, attacks, or saved results.
RESUME_FIX_SOURCE = r'''
"""Make temporary ZIP cleanup idempotent after a Colab runtime reset.

The original release and StegaStamp repair remain unchanged. Completed results
are verified first. Owner/checksum checks remain mandatory for every ZIP that
still exists; an absent ZIP requires no deletion and no owner file.
"""

import ast
import hashlib
from pathlib import Path

VERSION = "digital-missing-cache-v1"
STORAGE_SHA = "69f1f1362d337e8b955b8aee2e2da08e3a25d3cc62f6774c3de6c270bd2bf7de"


def install_cleanup_patch():
    """Move exactly one owner check inside the existing-file deletion branch."""
    from wmbench_digital import storage, runner

    source = Path(storage.__file__).read_bytes()
    if hashlib.sha256(source).hexdigest() != STORAGE_SHA:
        raise RuntimeError("Unexpected cleanup version; refusing to apply resume repair")
    tree = ast.parse(source.decode("utf-8"))
    function = next(n for n in tree.body if isinstance(n, ast.FunctionDef) and n.name == "remove_packages")
    loops = [n for n in function.body if isinstance(n, ast.For)
             and isinstance(n.target, ast.Name) and n.target.id == "manifest"]
    if len(loops) != 1:
        raise RuntimeError("Expected exactly one manifest cleanup loop")
    loop = loops[0]
    owner_test = ast.parse('read_json(folder / "owner.json") != {"identity": identity, "method": method}', mode="eval").body
    exists_test = ast.parse("file.exists()", mode="eval").body
    owner = [n for n in loop.body if isinstance(n, ast.If) and ast.dump(n.test) == ast.dump(owner_test)]
    deletion = [n for n in loop.body if isinstance(n, ast.If) and ast.dump(n.test) == ast.dump(exists_test)]
    if len(owner) != 1 or len(deletion) != 1:
        raise RuntimeError("Expected exactly one owner check and one conditional ZIP deletion")
    loop.body.remove(owner[0])
    deletion[0].body.insert(0, owner[0])
    module = ast.fix_missing_locations(ast.Module(body=[function], type_ignores=[]))
    exec(compile(module, str(storage.__file__) + " [missing-cache resume fix]", "exec"), storage.__dict__)
    runner.remove_packages = storage.remove_packages


def main():
    # Keep the previously supplied StegaStamp handling byte-for-byte unchanged.
    import stegastamp_clean_hotfix

    install_cleanup_patch()
    stegastamp_clean_hotfix.main()


if __name__ == "__main__":
    main()
'''
RESUME_FIX_SOURCE = RESUME_FIX_SOURCE.lstrip("\n")
RESUME_FIX_SHA = "94931da30679bd093a6f1790b4ea3058e869d50f9275f381e13e1101d22e0567"
if hashlib.sha256(RESUME_FIX_SOURCE.encode("utf-8")).hexdigest() != RESUME_FIX_SHA:
    raise ValueError("Resume repair source changed")
RESUME_FIX_DIR = SCRATCH_ROOT / "repairs/digital-missing-cache-v1"
for folder in (RESUME_FIX_DIR, OUTPUT_ROOT / "_release/repairs/digital-missing-cache-v1"):
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / "digital_resume_hotfix.py"
    if target.exists() and target.read_bytes() != RESUME_FIX_SOURCE.encode("utf-8"):
        raise ValueError(f"Different resume repair already exists: {target}")
    if not target.exists():
        target.write_bytes(RESUME_FIX_SOURCE.encode("utf-8"))
    if hashlib.sha256(target.read_bytes()).hexdigest() != RESUME_FIX_SHA:
        raise ValueError("Resume repair readback failed")
ENV["PYTHONPATH"] = str(RESUME_FIX_DIR) + os.pathsep + ENV["PYTHONPATH"]
print("Resume repair enabled: already absent temporary ZIPs need no cleanup.")
```

## StegaStamp CUDA runtime repair (supersedes the one-bit exception)

The full previous logs show `libcublasLt.so.12` missing. ONNX silently used CPU,
whereas the embedding bank records CUDA. The cleanup repair above is still used;
the earlier **one-bit Clean exception is NOT used** by this run entry point.

This overlay changes StegaStamp only:

1. Link the existing controller CUDA 12 / cuDNN 9 libraries into its subprocesses.
2. Require actual CUDA session providers; disable automatic CPU fallback.
3. Replay every selected Clean carrier, requiring exact agreement with the bank.
4. Only after all pass: preserve old, unverified-backend StegaStamp result files
   byte-for-byte under `stegastamp/_preserved_cpu_records/stegastamp-cuda-v1/` on
   Drive, then withdraw them from active tables and recompute their attacks.

The bank and the other methods' results are unchanged. Failed CUDA preflight
leaves previous active records in place and stops; no tolerance is increased.
Interrupted CUDA runs resume their verified partial records normally.

Keep the same bank/output paths and `N_IMAGES = 5`. Select the same NVIDIA GPU
as the bank (A100), then run all cells. The original source ZIP is reused from
Drive. No manual result deletion, payload re-embedding or new archive is needed.

Preflight evidence: `stegastamp/cuda-preflight/`. GPU-backed execution still
requires verification in Colab; local CPU smoke tests cannot establish bitwise
CUDA replay. Completed methods are reused, not decoded again.


```python
# StegaStamp only. No source archive, bank, attack or other decoder changes.
CUDA_FIX_SOURCE = r'''
"""Restore StegaStamp's CUDA runtime without changing the frozen benchmark.

The previous digital run silently used CPU because ONNX could not find CUDA
libraries. The embedding bank used CUDA. This overlay links the controller's
existing CUDA dependencies into StegaStamp children, verifies actual session
providers, and retains the original strict Clean gate. No weights, bits, images,
attacks, dependency locks or other methods are modified.
"""

import ast
import hashlib
import importlib.metadata
import json
import os
from pathlib import Path
import sys
import time

METHOD = "stegastamp"
VERSION = "stegastamp-cuda-v1"
RUNNER_SHA = "1357f717a1d46fbc3f726ecc097455241ac6d13e628b0f54b2741860febedf92"
WORKER_SHA = "57f68d4e9c6134c1f5264703e6e00ac1a6ef55a0040265f5b0c6dfd46ef6cf2e"
FIELDS = ("status", "decoded_bits", "ber", "exact")
REQUIRED_LIBRARIES = ("libcublasLt.so.12", "libcublas.so.12", "libcudnn.so.9", "libcudart.so.12")


def checksum(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as stream:
        for block in iter(lambda: stream.read(4 * 1024**2), b""):
            h.update(block)
    return h.hexdigest()


def library_environment(env):
    """Use existing PyTorch CUDA12/cuDNN9 wheels, outside the locked ORT venv."""
    names = ("cublas", "cuda_runtime", "cuda_nvrtc", "cudnn", "cufft", "curand",
             "cusolver", "cusparse", "nvjitlink", "nvtx")
    folders, packages = [], {}
    for name in names:
        package = "nvidia-" + name.replace("_", "-") + "-cu12"
        try:
            dist = importlib.metadata.distribution(package)
        except importlib.metadata.PackageNotFoundError:
            continue
        directory = Path(dist.locate_file("nvidia")) / name / "lib"
        if directory.is_dir():
            folders.append(directory.resolve())
            packages[package] = dist.version
    libraries = {}
    for filename in REQUIRED_LIBRARIES:
        matches = [folder / filename for folder in folders if (folder / filename).is_file()]
        if len(matches) != 1:
            raise RuntimeError(f"StegaStamp needs exactly one {filename} in the controller's NVIDIA wheels; "
                               "rerun the original environment setup cell. No results were removed.")
        path = matches[0].resolve()
        libraries[filename] = {"path": str(path), "sha256": checksum(path)}
    patched = dict(env)
    patched["LD_LIBRARY_PATH"] = os.pathsep.join([*map(str, folders), env.get("LD_LIBRARY_PATH", "")]).rstrip(os.pathsep)
    patched["WM_STEGA_CUDA_LIBRARIES"] = json.dumps({"packages": packages, "libraries": libraries}, sort_keys=True)
    return patched


def require_cuda(session):
    """Available providers are not evidence that this session loaded CUDA."""
    providers = session.get_providers()
    if not providers or providers[0] != "CUDAExecutionProvider":
        raise RuntimeError(f"StegaStamp CUDA did not load (actual providers: {providers}). "
                           "CPU-only fallback is forbidden; no attack scores accepted.")
    session.disable_fallback()  # Do not retry a failed CUDA session on CPU.
    return {"fix": VERSION, "actual_providers": providers,
            "provider_options": session.get_provider_options(), "automatic_fallback": False}


def install_worker_guard():
    """Wrap only StegaStamp's loader; retain the original decoding and Clean gate."""
    from wmbench_digital import worker
    from wmbench.adapters.physical import StegaStampONNX

    if checksum(worker.__file__) != WORKER_SHA:
        raise RuntimeError("Unexpected digital worker; refusing CUDA overlay")
    # Never compose this repair with the earlier one-bit acceptance exception.
    tree = ast.parse(Path(worker.__file__).read_text("utf-8"))
    decode = next(n for n in tree.body if isinstance(n, ast.FunctionDef) and n.name == "decode")
    exec(compile(ast.Module(body=[decode], type_ignores=[]), str(worker.__file__), "exec"), worker.__dict__)
    if getattr(StegaStampONNX._load, "_cuda_guard", False):
        return
    original_load, original_runtime = StegaStampONNX._load, worker.runtime_info
    evidence = {}

    def guarded_load(self):
        session = original_load(self)
        evidence.update(require_cuda(session))
        return session

    def runtime():
        return {**original_runtime(), "stegastamp_cuda": {
            **evidence, "linked_dependencies": json.loads(os.environ.get("WM_STEGA_CUDA_LIBRARIES", "{}"))}}

    guarded_load._cuda_guard = True
    StegaStampONNX._load = guarded_load
    worker.runtime_info = runtime


def receipt_path(job):
    return Path(job["results"]) / METHOD / "cuda-preflight" / (job["session"] + ".json")


def preflight(job):
    """All selected Clean carriers must replay exactly before replacing CPU rows."""
    from wmbench.io import read_json, write_json, sha256
    from wmbench_full_payload.storage import EvidenceError, inside
    from wmbench_digital import worker

    if job["model"]["id"] != METHOD:
        raise ValueError("CUDA preflight is StegaStamp-only")
    root, bank = Path(job["results"]), Path(job["bank"])
    cases = read_json(root / METHOD / "cases.json")
    install_worker_guard()
    adapter = worker.load_adapter(job)
    evidence = worker.runtime_info()
    print("STEGASTAMP ACTUAL PROVIDERS: " + str(evidence["stegastamp_cuda"]["actual_providers"]), flush=True)
    report = dict(fix=VERSION, session=job["session"], identity=job["identity"],
                  case_keys=[c["key"] for c in cases], execution=evidence, checks=[], status="running")
    write_json(receipt_path(job), report)
    for index, case in enumerate(cases, 1):
        if time.time() >= job["deadline"]:
            raise TimeoutError("Session budget reached during CUDA Clean preflight; CPU rows retained")
        carrier, metadata = inside(bank, case["output"]), inside(bank, case["decoder_metadata"])
        blind = read_json(metadata)
        if blind.get("runtime_providers", [None])[0] != "CUDAExecutionProvider":
            raise EvidenceError("Bank carrier was not recorded with CUDA; cannot silently change its backend")
        if any(any(blind.get(field, [])) for field in ("expected_bits", "payload", "full_payload")
               if isinstance(blind.get(field), list)):
            raise EvidenceError("Reference payload unexpectedly present in decoder metadata")
        if sha256(carrier) != case["carrier_sha256"]:
            raise EvidenceError("Bank carrier hash changed")
        fresh = worker.measured(adapter, carrier, metadata, case, int(case["signature"][:8], 16))
        passed = all(fresh.get(k) == case["clean"].get(k) for k in FIELDS)
        report["checks"].append({"case_key": case["key"], "case_id": case["id"],
                                  "fresh": fresh, "bank": case["clean"], "passed": passed})
        report["status"] = "running" if passed else "failed"
        write_json(receipt_path(job), report)
        print(f"STEGASTAMP CUDA CLEAN {index}/{len(cases)} {case['id']}: {'exact replay' if passed else 'FAILED'}", flush=True)
        if not passed:
            raise EvidenceError("StegaStamp still differs on actual CUDA; old results retained. "
                                "Inspect cuda-preflight; no tolerance or bit correction was applied.")
    report["status"] = "passed"
    write_json(receipt_path(job), report)


def cuda_record(document):
    """Only rows produced with the actual-provider guard are resumable as CUDA."""
    return bool(document and document["rows"] and all(
        row.get("execution", {}).get("runtime", {}).get("stegastamp_cuda", {}).get("fix") == VERSION
        and row["execution"]["runtime"]["stegastamp_cuda"].get("actual_providers", [None])[0] == "CUDAExecutionProvider"
        for row in document["rows"].values()))


def quarantine_legacy(job, cases):
    """Backup, verify and withdraw ONLY selected legacy StegaStamp record files.

    Content-addressed backups and append-only per-file receipts make a restart
    safe after any copy/write/unlink boundary. Bank PNGs and other methods never
    enter this function's target set. New CUDA rows are never withdrawn.
    """
    from wmbench.io import read_json, write_json, atomic_bytes
    from wmbench_digital.storage import result_path, read_result

    report = read_json(receipt_path(job))
    checked_keys = [row["case_key"] for row in report.get("checks", []) if row.get("passed") is True]
    runtime = report.get("execution", {}).get("stegastamp_cuda", {})
    if (job["model"]["id"] != METHOD or report["status"] != "passed"
            or report["fix"] != VERSION or report["session"] != job["session"]
            or report["identity"] != job["identity"]
            or runtime.get("fix") != VERSION
            or runtime.get("actual_providers", [None])[0] != "CUDAExecutionProvider"
            or checked_keys != [c["key"] for c in cases]
            or report["case_keys"] != [c["key"] for c in cases]):
        raise RuntimeError("Strict CUDA preflight must pass before legacy records can be archived")
    root = Path(job["results"]).resolve()
    active = root / METHOD / "records"
    backup_root = root / METHOD / "_preserved_cpu_records" / VERSION
    if any(p.is_symlink() for p in (root / METHOD, active, backup_root.parent, backup_root)):
        raise RuntimeError("Refusing linked result/backup directories")
    withdrawn = 0
    for case in cases:
        path = result_path(root, case)
        if case["method"] != METHOD or path.parent.resolve() != active.resolve() or path.is_symlink():
            raise RuntimeError("Unsafe legacy record target")
        document = read_result(path, case)
        if document is None or cuda_record(document):
            continue
        data = path.read_bytes()
        digest = hashlib.sha256(data).hexdigest()
        backup = backup_root / digest / path.name
        if backup.parent.is_symlink() or backup.is_symlink():
            raise RuntimeError("Refusing linked backup file")
        if backup.exists() and backup.read_bytes() != data:
            raise RuntimeError("Legacy backup conflict; active result retained")
        if not backup.exists():
            atomic_bytes(backup, data)
        if checksum(backup) != digest or read_result(backup, case) != document:
            raise RuntimeError("Legacy backup readback failed; active result retained")
        receipt = {"fix": VERSION, "reason": "legacy StegaStamp results without verified CUDA provider",
                   "case_key": case["key"], "source": str(path), "backup": str(backup),
                   "sha256": digest, "preflight": str(receipt_path(job)), "state": "backed_up"}
        receipt_file = backup.with_suffix(".receipt.json")
        write_json(receipt_file, receipt)
        if read_json(receipt_file) != receipt or checksum(path) != digest:
            raise RuntimeError("Record changed before withdrawal; active result retained")
        path.unlink()  # Exact checked record only; verified byte-identical backup persists on Drive.
        receipt["state"] = "withdrawn_from_active_tables"
        write_json(receipt_file, receipt)
        withdrawn += 1
    print(f"STEGASTAMP CUDA: {withdrawn} legacy records preserved under {backup_root}; "
          "only these records need recomputation. Other methods unchanged.", flush=True)


def redirect(command, env, linked_env):
    """Modify only StegaStamp prepare/decode children, never global LD paths."""
    from wmbench.io import read_json

    command = list(command)
    for index in range(len(command) - 4):
        if command[index:index+2] != ["-m", "wmbench_digital.runner"]:
            continue
        stage = command[index+2]
        if stage not in {"prepare", "decode"} or command[index+3] != "--settings":
            continue
        job = read_json(command[index+4])
        if job["model"]["id"] != METHOD:
            return command, env
        patched = dict(env)
        for key in ("LD_LIBRARY_PATH", "WM_STEGA_CUDA_LIBRARIES"):
            patched[key] = linked_env[key]
        if stage == "decode":
            command[index+1] = "stegastamp_cuda_hotfix"
            if job.get("stegastamp_cuda_preflight"):
                command[index+2] = "preflight"
            patched["PYTHONPATH"] = str(Path(__file__).resolve().parent) + os.pathsep + env.get("PYTHONPATH", "")
        return command, patched
    return command, env


def install_controller_patch():
    """Insert one guarded StegaStamp preparation hook before original stages."""
    from wmbench import colab_canary
    from wmbench.io import read_json, write_json
    from wmbench_digital import runner
    from wmbench_digital.storage import read_result, result_path, complete
    from digital_resume_hotfix import install_cleanup_patch

    install_cleanup_patch()
    if checksum(runner.__file__) != RUNNER_SHA:
        raise RuntimeError("Unexpected controller version; refusing CUDA overlay")
    original_monitor, original_invoke = colab_canary.monitored_process, runner.invoke
    prepared_sessions, linked = set(), {}

    def monitored(command, env, *args, **kwargs):
        # Lazy: completed methods need no CUDA preparation or linkage changes.
        if linked:
            command, env = redirect(command, env, linked)
        return original_monitor(command, env, *args, **kwargs)

    def invoke(job, stage, path, owned):
        if job["model"]["id"] == METHOD and stage == "prepare" and job["session"] in prepared_sessions:
            return
        return original_invoke(job, stage, path, owned)

    def before(job, path, owned, cases):
        if job["model"]["id"] != METHOD:
            return
        documents = [read_result(result_path(job["results"], c), c) for c in cases]
        if all(complete(d) and cuda_record(d) for d in documents):
            return
        linked.update(library_environment(os.environ))
        owned()
        original_invoke(job, "prepare", path, owned)
        prepared_sessions.add(job["session"])
        probe = {**job, "session": job["session"] + "_cuda-preflight", "stegastamp_cuda_preflight": True}
        probe_path = Path(path).with_name(probe["session"] + ".json")
        write_json(probe_path, probe)
        original_invoke(probe, "decode", probe_path, owned)
        owned()
        quarantine_legacy(probe, cases)
        # Original attacks stage now sees archived cases as pending and recreates
        # missing ZIPs with its existing deterministic SHA checks.

    tree = ast.parse(Path(runner.__file__).read_text("utf-8"))
    function = next(n for n in tree.body if isinstance(n, ast.FunctionDef) and n.name == "run")
    count = 0

    class InsertBeforeDone(ast.NodeTransformer):
        def visit_Assign(self, node):
            nonlocal count
            if len(node.targets) == 1 and isinstance(node.targets[0], ast.Name) and node.targets[0].id == "done":
                count += 1
                return [ast.copy_location(ast.parse("_stega_cuda_before(job, path, owned, cases)").body[0], node), node]
            return self.generic_visit(node)

    function = InsertBeforeDone().visit(function)
    if count != 1:
        raise RuntimeError("Expected exactly one method-completion boundary")
    runner.__dict__["_stega_cuda_before"] = before
    exec(compile(ast.fix_missing_locations(ast.Module(body=[function], type_ignores=[])),
                 str(runner.__file__) + " [StegaStamp CUDA linkage]", "exec"), runner.__dict__)
    colab_canary.monitored_process = monitored
    runner.invoke = invoke


def main():
    from wmbench.io import read_json
    from wmbench_digital import runner, worker

    action = sys.argv[1] if len(sys.argv) > 1 else ""
    if action == "run":
        install_controller_patch()
        runner.main()
    elif action in {"decode", "preflight"}:
        job = read_json(Path(sys.argv[sys.argv.index("--settings")+1]))
        if job["model"]["id"] != METHOD:
            raise ValueError("CUDA repair decoder is StegaStamp-only")
        if action == "preflight":
            preflight(job)
        else:
            install_worker_guard()
            worker.decode(job)
    else:
        raise ValueError("Use original runner for validation/report; this overlay supports run/decode/preflight")


if __name__ == "__main__":
    main()
'''
CUDA_FIX_SOURCE = CUDA_FIX_SOURCE.lstrip("\n")
CUDA_FIX_SHA = "05ec6a874215806f299cca507e5573514a25014adb731ffd9d53e09bf08fb7aa"
if hashlib.sha256(CUDA_FIX_SOURCE.encode()).hexdigest() != CUDA_FIX_SHA:
    raise ValueError("CUDA repair source changed")
CUDA_FIX_DIR = SCRATCH_ROOT / "repairs/stegastamp-cuda-v1"
for folder in (CUDA_FIX_DIR, OUTPUT_ROOT / "_release/repairs/stegastamp-cuda-v1"):
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / "stegastamp_cuda_hotfix.py"
    data = CUDA_FIX_SOURCE.encode("utf-8")
    if target.exists() and target.read_bytes() != data:
        raise ValueError(f"Different CUDA repair already exists: {target}")
    if not target.exists():
        target.write_bytes(data)
    if hashlib.sha256(target.read_bytes()).hexdigest() != CUDA_FIX_SHA:
        raise ValueError("CUDA repair readback failed")
ENV["PYTHONPATH"] = str(CUDA_FIX_DIR) + os.pathsep + ENV["PYTHONPATH"]
print("StegaStamp: require actual CUDA, strict Clean replay, preserve legacy CPU results before recomputation.")
```

```python
# Stop any previous runner for this output folder before starting.
# After a disconnect rerun the same cells with the same bank/output and source ZIP.
command = [str(CONTROLLER), "-m", "stegastamp_cuda_hotfix", "run", "--settings", str(CONFIG)]
process = subprocess.Popen(command, env=ENV, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
    text=True, encoding="utf-8", errors="replace", start_new_session=True)
try:
    for line in process.stdout:
        print(line, end="", flush=True)
    if process.wait():
        raise RuntimeError("Runner stopped. See persistent per-method logs; completed records remain safe.")
except BaseException:
    if process.poll() is None:
        import signal
        os.killpg(process.pid, signal.SIGTERM)
        process.wait()
    raise
```
